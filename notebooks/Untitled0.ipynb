### Simple Login Page with Password Validator

This cell implements a basic, interactive login interface using `ipywidgets`. It includes fields for username and password, along with a validator that checks if the password meets basic security standards (e.g., minimum length of 6 characters).

````python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create input widgets
username_input = widgets.Text(
    value='',
    placeholder='Enter your username',
    description='Username:',
    disabled=False
)

password_input = widgets.Password(
    value='',
    placeholder='Enter password',
    description='Password:',
    disabled=False
)

login_button = widgets.Button(
    description='Log In',
    disabled=False,
    button_style='primary',
    tooltip='Click to login'
)

output = widgets.Output()

def validate_password(password):
    # Basic validation: check if length is at least 6 characters
    if len(password) < 6:
        return False, "Password must be at least 6 characters long."
    if not any(char.isdigit() for char in password):
        return False, "Password must contain at least one number."
    return True, "Valid password."

def on_login_clicked(b):
    with output:
        clear_output()
        username = username_input.value.strip()
        password = password_input.value

        if not username:
            print("Error: Username cannot be empty.")
            return

        is_valid, message = validate_password(password)
        if is_valid:
            print(f"Success! Welcome, {username}.")
        else:
            print(f"Login Failed: {message}")

login_button.on_click(on_login_clicked)

# Display the form
login_form = widgets.VBox([username_input, password_input, login_button, output])
display(login_form)
````

## How to Deploy on Localhost

To run this interactive login page on your local computer, you can choose one of the following methods:

### Method 1: Using Jupyter Notebook or JupyterLab

If you see an error saying `'jupyter' is not recognized as an internal or external command`, it means python's script folder isn't in your Windows PATH. You can bypass this by running Python directly.

1. **Install Dependencies**: Open your PowerShell/Command Prompt and run:
   ```bash
   pip install jupyterlab ipywidgets
   ```

2. **Launch Jupyter (Bypassing PATH issues)**:
   If typing `jupyter lab` fails, use `python -m` to run it directly:
   ```bash
   python -m jupyterlab
   ```
   *(Alternatively, if you prefer the classic notebook interface: `python -m notebook`)*

3. **Note on Widgets**:
   If you installed modern JupyterLab or Jupyter Notebook (v7+), you **do not** need to run `jupyter nbextension enable`. The widgets will work automatically out of the box!

---

### Method 2: Deploy as a Standalone Web App using Voila

If you want to hide the raw code and present the login interface as a clean web page, you can use **Voila**:

1. **Install Voila**:
   ```bash
   pip install voila ipywidgets
   ```
2. **Save your Code**: Save the login python code in a notebook file named `login_app.ipynb`.
3. **Run Voila**:
   If `voila` is not recognized as a command, run it via Python:
   ```bash
   python -m voila login_app.ipynb
   ```
4. **Access on Localhost**: Your web browser will open `http://localhost:8866/` showing only the modern, styled login container.
